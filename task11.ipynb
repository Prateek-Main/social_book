```python
import numpy as np
```

```python
import pandas as pd
```

```python
data = {
    'Column1': np.random.randint(1, 100, 10),
    'Column2': np.random.randint(1, 100, 10),
    'Column3': np.random.randint(1, 100, 10),
}
```

```python
df = pd.DataFrame(data)
print("DataFrame:")
print(df)
```

```python
# Filter rows where 'Column1' > 50
filtered_df = df[df['Column1'] > 50]
print("\nFiltered DataFrame (Column1 > 50):")
print(filtered_df)
```

```python
# Selecting two columns
filtered_columns = df[['Column1', 'Column2']]
print("\nFiltered DataFrame (Only Column1 and Column2):")
print(filtered_columns)
```

```python
# Replace values in 'Column1' greater than 80 with 999
df.loc[df['Column1'] > 80, 'Column1'] = 999
print("\nDataFrame after replacing values in Column1 > 80 with 999:")
print(df)
```

```python
# Creating another DataFrame with the same columns
data2 = {
    'Column1': np.random.randint(1, 100, 10),
    'Column2': np.random.randint(1, 100, 10),
    'Column3': np.random.randint(1, 100, 10),
}
df2 = pd.DataFrame(data2)

# Appending the DataFrames
appended_df = pd.concat([df, df2], ignore_index=True)
print("\nAppended DataFrame:")
print(appended_df)
```
